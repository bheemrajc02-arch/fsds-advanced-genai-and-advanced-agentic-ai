```python
# .format method  usage
# With .format method usage
# In .format method the number of variables is equals to number of curley bases

# .format method the number of variables = number of curly braces user must need to mentio

num1 = 20
num2 = 30
add = num1+num2
print('The addition of {} and {} is {}'.format(num1,num2,add))
```

```python
# .format method  usage

# With out .format method
num1= 20
num2 = 30
num3 = 50
add = num1+num2+num3
print('The addition of', num1, 'and', num2, 'and', num3, 'is =', add)
```

```python
# .format method  usage
# With .format method usage
# In .format method the number of variables is equals to number of curley bases

# .format method the number of variables = number of curly braces user must need to mention

# most working proffessional use this code

num1 = 20
num2 = 30
num3 = 40
add = num1+num2+num3
print('The addition of {} and {} and {}  is {}'.format(num1,num2,num3,add))
```

```python
# f string method 
# more short format method is f string method

num1 = 30
num2 = 40
add = num1+num2
# We need to write short format in print as f
print(f'The addition of {num1} and {num2} is = {add}')     # We need to prefer always this
```
